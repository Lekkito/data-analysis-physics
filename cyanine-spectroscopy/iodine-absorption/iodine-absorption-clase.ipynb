# Absorción de I$_2$: niveles vibracionales — Laboratorio 3

Espectro de absorción de I$_2$ gaseoso (Shimadzu UV-1650, 500–650 nm). Cada pico es una transición X($v''$) → B($v'$).

# Librerías

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path
```

# Importando datos

```python
# El archivo trae la transmitancia en % y usa coma decimal
df = pd.read_csv(
    Path("datos_I2") / "2025" / "DATOS_I2_2025.dat",
    sep="\t", decimal=",", header=0, names=["lambda", "T"],
)

# Absorbancia: A = log10(I0/I) = -log10(T/100)
df["A"] = -np.log10(df["T"] / 100)

df.head()
```

# Primer gráfico: absorbancia vs longitud de onda

```python
fig, ax = plt.subplots(figsize=(10, 4.5))
ax.plot(df["lambda"], df["A"])
ax.set_xlabel(r"$\lambda$ (nm)")
ax.set_ylabel("Absorbancia")
ax.set_title("Espectro de absorción de I$_2$")
ax.grid(True)
plt.show()
```
